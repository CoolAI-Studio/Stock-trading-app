# 策略回測報告 (中文)
這份報告展示了每日回測的績效指標，包括 **夏普比率**、**最大回撤**、以及盈虧分布。

```python
import matplotlib.pyplot as plt
import numpy as np
from src.backtesting import Backtester
from src.database_manager import DatabaseManager

# 從資料庫讀取回測結果
db = DatabaseManager("trading_app.db")
results = db.get_backtest_results()
print("回測結果:", results)

# 模擬回測數據
returns = [0.01, -0.02, 0.03, -0.01, 0.02]
bt = Backtester(returns)
print("夏普比率:", bt.sharpe_ratio())
print("最大回撤:", bt.max_drawdown())
```

```python
# 繪製盈虧分布圖
plt.hist(returns, bins=5, edgecolor='black')
plt.title("盈虧分布圖")
plt.xlabel("每日報酬率")
plt.ylabel("次數")
plt.show()
```
